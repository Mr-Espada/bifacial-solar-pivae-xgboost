```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_train = pd.read_csv("../train_simulation.csv")
df_test = pd.read_csv("../test_simulation.csv")

df = pd.concat([df_train, df_test])
df.head()
```

```python
std_Q = np.std(df["Q_1kW"], axis=0)
mu_Q = np.mean(df["Q_1kW"], axis=0)
```

```python
# reorder based on Datum column
df["Datum"]  = pd.to_datetime(df["Datum"], format="%Y-%m-%d %H:%M:%S")
df = df.sort_values(by=['Datum'])
```

```python
# plot Q
plt.figure(figsize=(20, 5))
plt.plot(df["Datum"], df["T_out_1C"])
# remove paddings
plt.xlim(df["Datum"].min(), df["Datum"].max())
```

```python
plt.plot(df["Datum"][-10000:], df["Q_1kW"][-10000:])
```

```python
#  Z score scaling of the data
from scipy.stats import zscore

dfz = df.copy()

dfz.drop(columns=["Datum"], inplace=True)

dfz = dfz.apply(zscore)
```

```python
plt.figure(figsize=(20, 5))
plt.plot(df["Datum"], df["m_flow_1kgh"])
```

```python
posterior = result.mean().to_numpy()
```

```python
df_test["Datum"] = pd.to_datetime(df_test['Datum'])
```

```python
plt.plot(df_test["Datum"], posterior)
```

```python
plt.plot(df_test["Datum"], df_test["Q_1kW"])
```

```python
posterior_raw = posterior * std_Q + mu_Q
```

```python
plt.rcParams['figure.dpi'] = 300

plt.plot(df_test["Datum"], posterior_raw, label="Posterior", alpha=1)

# Plot the second line on the same figure
plt.plot(df_test["Datum"], df_test["Q_1kW"], label="Q_1kW", alpha=0.5)

# Adding labels and title
plt.xlabel("Datum")
plt.ylabel("Values")
plt.title("Posterior vs Q_1kW")

# Display a legend to differentiate the plots
plt.legend()

# Show the plot
plt.show()
```
